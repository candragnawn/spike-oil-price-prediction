## 3. Data Cleansing & Transformasi Struktur (Wide to Long)

Dataset mentah dari PIHPS berformat **wide** (setiap kolom mewakili tanggal pengamatan tertentu dalam format string `DD/ MM/ YYYY`).  
Format ini harus ditransformasikan ke format **long** terstruktur untuk memenuhi kaidah analisis deret waktu dan persiapan input model.

Langkah pembersihan data:
1. **Unpivot / Melting**: Mengubah kolom-kolom tanggal menjadi baris individual.
2. **Standardisasi Tanggal**: Menghapus spasi acak pada teks tanggal dan mengubahnya ke objek `datetime` dengan format `%d/%m/%Y`.
3. **Pembersihan Angka Nominal**: Menghapus tanda pemisah ribuan (koma `,`) dan mengonversi tanda ketiadaan data (`-`) menjadi `NaN` bertipe `float`.
4. **Ekstraksi Komponen Waktu**: Menambahkan fitur `Tahun`, `Bulan`, `Bulan_Nama`, dan `Minggu_ke` untuk analisis musiman.


```python
def bersihkan_data(df_wide):
    """
    Mentransformasikan dataframe PIHPS format wide menjadi format long terstandarisasi.
    """
    nama_kolom_komoditas = df_wide.columns[1]  # Kolom ke-2 = Nama Komoditas
    
    # Identifikasi kolom tanggal
    kolom_tanggal = [
        col for col in df_wide.columns
        if col not in ['No', nama_kolom_komoditas, 'Komoditas', 'Komoditas (Rp)']
    ]
    
    # Melt: wide -> long
    df_long = pd.melt(
        df_wide,
        id_vars=[col for col in df_wide.columns if col not in kolom_tanggal],
        value_vars=kolom_tanggal,
        var_name='Tanggal_Str',
        value_name='Harga_Raw',
    )
    
    # Standarisasi nama kolom
    df_long = df_long.rename(columns={nama_kolom_komoditas: 'Komoditas'})
    
    # Parsing datetime
    df_long['Tanggal'] = pd.to_datetime(
        df_long['Tanggal_Str'].astype(str).str.replace(' ', ''),
        format='%d/%m/%Y',
        errors='coerce'
    )
    
    # Konversi harga numerik (hilangkan tanda koma dan strip)
    df_long['Harga'] = pd.to_numeric(
        df_long['Harga_Raw'].astype(str).str.replace(',', '').str.replace('-', ''),
        errors='coerce'
    )
    
    # Fitur kalender
    df_long['Tahun'] = df_long['Tanggal'].dt.year
    df_long['Bulan'] = df_long['Tanggal'].dt.month
    df_long['Bulan_Nama'] = df_long['Tanggal'].dt.strftime('%B')
    df_long['Minggu_ke'] = df_long['Tanggal'].dt.isocalendar().week.astype(int)
    
    # Drop kolom penolong
    df_long = df_long.drop(columns=['Tanggal_Str', 'Harga_Raw'])
    
    # Urutkan secara kronologis per komoditas
    df_long = df_long.sort_values(['Komoditas', 'Tanggal']).reset_index(drop=True)
    return df_long

# Eksekusi pembersihan
df_long = bersihkan_data(df_raw)

print(f"Data Long berhasil dibentuk!")
print(f"  - Total Observasi: {len(df_long):,} baris")
print(f"  - Rentang Waktu: {df_long['Tanggal'].min().date()} s/d {df_long['Tanggal'].max().date()}")
print(f"  - Komoditas ({df_long['Komoditas'].nunique()} varian): {df_long['Komoditas'].unique().tolist()}")

display(Markdown("### Pratinjau 10 Baris Pertama Data Bersih"))
display(df_long.head(10))
```

### Strategi Penanganan Ganda (Dual Pipeline Strategy)

Sesuai metodologi pada `project_context.md`, penanganan missing values dibagi menjadi dua jalur independen:
1. **Jalur Pemodelan Prophet (Model Utama):**  
   Baris dengan nilai kosong dihilangkan (`dropna()`). Prophet memiliki arsitektur *continuous-time generative model* yang secara native mampu menangani jarak waktu diskontinu (*irregular time intervals*) tanpa memerlukan imputasi buatan yang dapat mendistorsi slope tren.
2. **Jalur EDA & Supervised ML Regresi (Model Sekunder / Topik 7):**  
   Diberlakukan **interpolasi linear teratur** (`interpolate(method='linear')`). Pendekatan ini mempertahankan kontinuitas kronologis pekan-ke-pekan agar pembentukan fitur lag autoregresif ($y_{t-1}, y_{t-2}$) tidak terputus.


```python
print("=== AUDIT KEKOSONGAN DATA (MISSING VALUES) ===")

# Rekapitulasi missing per varian dan per tahun
tabel_missing = (
    df_long.groupby(['Komoditas', 'Tahun'])['Harga']
    .apply(lambda x: x.isna().sum())
    .unstack(fill_value=0)
)
display(Markdown("#### Tabel Jumlah Data Hilang per Komoditas per Tahun"))
display(tabel_missing)

# Tanggal eksak data hilang
missing_dates = df_long[df_long['Harga'].isna()][['Komoditas', 'Tanggal']].drop_duplicates()
display(Markdown("#### Tanggal Spesifik dengan Data Hilang"))
display(missing_dates.reset_index(drop=True))
```

```python
# 1. Jalur Prophet: Dataset format ds (tanggal) dan y (harga)
VARIAN_LIST = [
    'Minyak Goreng',
    'Minyak Goreng Curah',
    'Minyak Goreng Kemasan Bermerk 1',
    'Minyak Goreng Kemasan Bermerk 2',
]

datasets_prophet = {}
for varian in VARIAN_LIST:
    df_v = (
        df_long[df_long['Komoditas'] == varian][['Tanggal', 'Harga']]
        .rename(columns={'Tanggal': 'ds', 'Harga': 'y'})
        .dropna()  # Hapus baris NaN secara elegan untuk Prophet
        .sort_values('ds')
        .reset_index(drop=True)
    )
    datasets_prophet[varian] = df_v

print("Jalur 1 (Prophet): Dataset per varian siap.")
for varian, d_pr in datasets_prophet.items():
    print(f"  - {varian:32s}: {len(d_pr)} baris | Rentang: {d_pr['ds'].min().date()} s/d {d_pr['ds'].max().date()}")

# 2. Jalur EDA & Supervised ML: Imputasi Interpolasi Linear
df_eda = df_long.copy()
for varian in VARIAN_LIST:
    mask = df_eda['Komoditas'] == varian
    df_eda.loc[mask, 'Harga'] = (
        df_eda.loc[mask, 'Harga']
        .interpolate(method='linear', limit_direction='both')
    )

sisa_na = df_eda['Harga'].isna().sum()
print()
print(f"Jalur 2 (EDA & Supervised ML): Dataset terinterpolasi siap! (Sisa NaN: {sisa_na})")
```

## 5. Deteksi & Perbaikan Noise Data (Noise Tipe 2: Transient Noise vs Macroeconomic Shocks)

### Konseptual: Membedakan Noise Teknis vs Gejolak Harga Riil
Dalam analisis time series komoditas pangan, pergerakan data tersusun dari:
$$\text{Harga}(t) = \text{Tren Makroekonomi}(t) + \text{Efek Musiman \& Hari Libur}(t) + \text{Gejolak Riil}(t) + \text{Noise Acak}(t)$$

Sangat penting untuk membedakan dua jenis fluktuasi tajam:
1. **Transient High-Frequency Noise (Noise Acak):** Fluktuasi kecil mingguan (biasanya < 1–2%) yang disebabkan oleh dinamika pencatatan harian di pasar tradisional.
2. **Macroeconomic Price Shock (Gejolak Harga Riil):** Lonjakan harga tajam dan persisten yang terjadi akibat faktor fundamental (misalnya: krisis minyak goreng Maret–Mei 2022 pasca penghapusan kebijakan HET dan embargo ekspor CPO).

> **Catatan Penting:**  
> Jika lonjakan ekstrem pada kuartal awal 2022 dihapus atau dihaluskan secara agresif atas dalih "outlier removal", sistem akan kehilangan sinyal target utama yang justru ingin dideteksi oleh sistem peringatan dini (*early warning system*). Oleh karena itu, kita melakukan **Hampel Filter (Rolling Median Absolute Deviation)** untuk mendeteksi kandidat anomali, lalu memverifikasi apakah kandidat tersebut merupakan noise atau gejolak riil yang wajib dipertahankan.


```python
def hampel_filter(series, window_size=7, n_sigmas=3):
    """
    Deteksi pencilan lokal menggunakan Rolling Median dan Rolling MAD (Median Absolute Deviation).
    """
    rolling_median = series.rolling(window=window_size, center=True).median()
    rolling_mad = (series - rolling_median).abs().rolling(window=window_size, center=True).median()
    threshold = n_sigmas * 1.4826 * rolling_mad
    difference = (series - rolling_median).abs()
    outlier_idx = difference > threshold
    return outlier_idx, rolling_median

fig, axes = plt.subplots(2, 2, figsize=(16, 10), sharex=True)
fig.suptitle('Deteksi Kandidat Anomali & Noise vs Tren Fundamental (Hampel Filter)', fontsize=15, fontweight='bold')

anomali_summary = []

for ax, varian in zip(axes.flat, VARIAN_LIST):
    data_v = df_eda[df_eda['Komoditas'] == varian].sort_values('Tanggal').set_index('Tanggal')
    s = data_v['Harga']
    
    is_outlier, r_median = hampel_filter(s, window_size=7, n_sigmas=3)
    
    # Plot kurva harga dan tren rolling median
    ax.plot(s.index, s.values, label='Harga Aktual', color=COLORS[varian], linewidth=1.8, alpha=0.85)
    ax.plot(r_median.index, r_median.values, label='Tren Rolling Median (Filter Sinyal)',
            color='#264653', linestyle='--', linewidth=1.5)
    
    # Sorot titik-titik anomali/lonjakan
    if is_outlier.sum() > 0:
        ax.scatter(s.index[is_outlier], s[is_outlier], color='red', s=40, zorder=5,
                   label=f'Kandidat Anomali ({is_outlier.sum()} titik)')
    
    ax.set_title(f"{varian}", fontsize=12, fontweight='bold')
    ax.set_ylabel("Harga (Rp)")
    ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, p: f'Rp {x:,.0f}'))
    ax.grid(True, alpha=0.3)
    ax.legend(loc='upper left', fontsize=9)
    
    anomali_summary.append({
        'Varian': varian,
        'Jumlah Titik Deviasi Ekstrem': is_outlier.sum(),
        'Persentase terhadap Total Observasi': f"{(is_outlier.sum() / len(s) * 100):.1f}%",
        'Periode Kluster Utama': "Maret - Mei 2022 (Krisis Minyak Goreng)"
    })

plt.tight_layout()
plt.show()

display(Markdown("### Ringkasan Deteksi Deviasi Ekstrem (Kandidat Shock vs Noise)"))
display(pd.DataFrame(anomali_summary))

display(Markdown("""
**Kesimpulan Evaluasi Noise:**
1. Titik-titik deviasi ekstrem terkonsentrasi secara teratur pada rentang **Maret – Mei 2022**.
2. Ini membuktikan bahwa anomali tersebut **BUKAN noise kesalahan pencatatan**, melainkan peristiwa historis riil (krisis pasokan & restrukturisasi kebijakan minyak goreng 2022).
3. **Keputusan Data Preparation:** Titik-titik ini **wajib dipertahankan** dalam dataset agar model dapat mempelajari karakteristik gejolak harga ekstrem untuk fungsi sistem peringatan dini.
"""))
```

## 6. Kalender Hari Besar Keagamaan (External Regressor Calendar)

Model Prophet memiliki keunggulan utama dalam mengakomodasi efek hari libur (*holidays*) secara eksplisit.  
Hari raya keagamaan di Indonesia (seperti Idul Fitri, Idul Adha, Nyepi, dan Imlek) bergeser setiap tahun mengikuti kalender lunar (Hijriah, Saka, dan Imlek).

### Kalibrasi Jendela Waktu (Windowing Strategy):
- **Hari Raya Berdampak Sangat Tinggi (Idul Fitri):** Diberi `lower_window = -14` dan `upper_window = 7`. Ini karena lonjakan belanja pangan rumah tangga telah terjadi sejak 2 pekan sebelum hari raya (fase Ramadan aktif) dan berlanjut hingga 1 pekan setelah Lebaran.

## 7. Exploratory Data Analysis (EDA) Komprehensif

Tujuan tahapan EDA:
1. Meneliti sebaran statistik dasar dan tingkat volatilitas harga tiap varian.
2. Menganalisis tren jangka panjang (2022–2026) dan mendeteksi titik pergantian rezim harga.
3. Mengidentifikasi pola musiman bulanan (*monthly seasonality*).
4. Menganalisis korelasi dan dependensi pergerakan harga antarvarian minyak goreng.
5. Melakukan **Event Study** untuk memeriksa perilaku harga pada jendela waktu relatif ($H-28$ s/d $H+28$ hari) sekitar hari raya keagamaan.


```python
# 7.1 Statistik Deskriptif Harga Mingguan
statistik = (
    df_long.groupby('Komoditas')['Harga']
    .agg(
        N='count',
        Data_Hilang=lambda x: x.isna().sum(),
        Min='min',
        Maks='max',
        Rata_rata='mean',
        Median='median',
        Std_Dev='std',
    )
    .reset_index()
)

# Koefisien Variasi (CV) dalam %
statistik['CV_persen'] = (statistik['Std_Dev'] / statistik['Rata_rata'] * 100).round(2)

for col in ['Min', 'Maks', 'Rata_rata', 'Median', 'Std_Dev']:
    statistik[col] = statistik[col].round(0)

statistik.columns = [
    'Varian Minyak Goreng', 'N Valid', 'Data Hilang',
    'Min (Rp)', 'Maks (Rp)', 'Rata-rata (Rp)', 'Median (Rp)',
    'Std. Dev (Rp)', 'CV (%)'
]

display(Markdown("### Tabel Statistik Deskriptif Harga Mingguan Nasional (2022 - 2026)"))
display(statistik)

display(Markdown("""
> **Insight Statistik:**  
> - **Minyak Goreng Curah** memiliki koefisien variasi tertinggi (**10.35%**) dan standar deviasi terbesar (**Rp 1.796**). Ini mencerminkan karakteristik komoditas curah yang paling sensitif terhadap fluktuasi harga bahan baku CPO global dan perubahan kebijakan subsidi domestik.
> - **Minyak Goreng Kemasan Bermerk 1** memiliki harga rata-rata tertinggi (**Rp 22.144**), mencerminkan premi merek dan kemasan premium.
"""))
```

## 8. Feature Engineering untuk Supervised Machine Learning (Pencegahan Noise Tipe 3: Data Leakage)

Untuk membangun model prediksi berbasis regresi tabular atau komparasi machine learning, kita merekayasa fitur-fitur prediktif dari data historis.

### Prinsip Pencegahan Data Leakage (Kebocoran Informasi)
Dalam peramalan deret waktu mingguan:
- **Target yang diprediksi:** Harga minyak goreng pada waktu $t$ (`Harga_MG`).
- **Aturan Ketat:** Penggunaan harga varian lain (misalnya Curah dan Kemasan) pada waktu $t$ yang sama adalah bentuk **kebocoran data (data leakage)** karena data waktu $t$ belum terjadi saat peramalan dilakukan.
- **Penyelesaian:** Seluruh harga varian pendukung wajib digeser satu pekan ke belakang ($t-1$): `Curah_Lag_1`, `Kemasan1_Lag_1`, `Kemasan2_Lag_1`.

### Fitur-Fitur yang Direkayasa:
1. **Fitur Lag Target:** `Lag_1`, `Lag_2`, `Lag_3`, `Lag_4` (nilai harga 1 s/d 4 pekan sebelumnya).
2. **Fitur Cross-Variant Lag:** `Curah_Lag_1`, `Kemasan1_Lag_1`, `Kemasan2_Lag_1` (nilai varian substitusi pada $t-1$).
3. **Fitur Rolling & Volatilitas:** `Roll_Mean_4` (rata-rata 4 pekan terakhir), `Roll_Std_4` (volatilitas 4 pekan terakhir), dan `Pct_Change_1` (persentase perubahan pekan lalu).
4. **Fitur Kalender Siklikal:** `Bulan`, `Minggu_ke`, `Tahun`.
5. **Fitur Biner Hari Besar Keagamaan:** `Ramadan`, `Idul_Fitri`, `Idul_Adha`, `Nyepi`, `Natal` (bernilai 1 jika rentang pekan mencakup jendela perayaan).


```python
# Membangun matriks harga mingguan terinterpolasi
TARGET = 'Harga_MG'
raw_wide = df_raw.drop(columns='No').set_index(df_raw.columns[1]).T
raw_wide.index = pd.to_datetime(raw_wide.index.str.replace(' ', ''), format='%d/%m/%Y')
raw_wide = raw_wide.apply(lambda c: pd.to_numeric(c.astype(str).str.replace(',', '').replace('-', np.nan)))
raw_wide.columns = ['Harga_MG', 'Harga_Curah', 'Harga_Kemasan1', 'Harga_Kemasan2']
harga_ts = raw_wide.interpolate(method='linear')

s_target = harga_ts[TARGET]
fitur = pd.DataFrame(index=harga_ts.index)

# 1. Autoregressive Lags Target
fitur['Lag_1'] = s_target.shift(1)
fitur['Lag_2'] = s_target.shift(2)
fitur['Lag_3'] = s_target.shift(3)
fitur['Lag_4'] = s_target.shift(4)

# 2. Cross-Variant Lags (t-1) - Mencegah Data Leakage
fitur['Curah_Lag_1']    = harga_ts['Harga_Curah'].shift(1)
fitur['Kemasan1_Lag_1'] = harga_ts['Harga_Kemasan1'].shift(1)
fitur['Kemasan2_Lag_1'] = harga_ts['Harga_Kemasan2'].shift(1)

# 3. Rolling Statistics & Volatilitas
fitur['Roll_Mean_4']  = s_target.shift(1).rolling(4).mean()
fitur['Roll_Std_4']   = s_target.shift(1).rolling(4).std()
fitur['Pct_Change_1'] = s_target.shift(1).pct_change()

# 4. Kalender Waktu
fitur['Bulan']     = fitur.index.month
fitur['Minggu_ke'] = fitur.index.isocalendar().week.astype(int)
fitur['Tahun']     = fitur.index.year

# 5. Fitur Indikator Hari Besar Keagamaan
libur_dict = {
    'Ramadan':    (['2022-05-02', '2023-04-22', '2024-04-10', '2025-03-31', '2026-03-21'], -30, -1),
    'Idul_Fitri': (['2022-05-02', '2023-04-22', '2024-04-10', '2025-03-31', '2026-03-21'], -7, 7),
    'Idul_Adha':  (['2022-07-09', '2023-06-29', '2024-06-17', '2025-06-06', '2026-05-27'], -7, 3),
    'Nyepi':      (['2022-03-03', '2023-03-22', '2024-03-11', '2025-03-29', '2026-03-19'], -3, 1),
    'Natal':      (['2022-12-25', '2023-12-25', '2024-12-25', '2025-12-25', '2026-12-25'], -7, 3),
}

pekan_akhir = fitur.index + pd.Timedelta(days=6)
for nama, (tanggal, bawah, atas) in libur_dict.items():
    fitur[nama] = 0
    for d in pd.to_datetime(tanggal):
        fitur[nama] |= ((fitur.index <= d + pd.Timedelta(days=atas)) &
                        (pekan_akhir >= d + pd.Timedelta(days=bawah))).astype(int)

# Hapus baris NaN akibat shifting / lagging awal
df_ml = fitur.assign(Harga_MG=s_target).dropna()
X = df_ml.drop(columns='Harga_MG')
y = df_ml['Harga_MG']

print("Matriks fitur Supervised ML siap.")
print(f"  - Dimensi: {X.shape[0]} observasi mingguan x {X.shape[1]} fitur prediktor")
print(f"  - Rentang Data Latih: {X.index.min().date()} s/d {X.index.max().date()}")
print(f"  - Target Prediksi: '{TARGET}' (Min: Rp {y.min():,.0f}, Maks: Rp {y.max():,.0f})")
```

## 9. Seleksi Fitur Multi-Metode (Pencegahan Noise Tipe 4: Redundansi & Multikolinearitas)

Menggunakan terlalu banyak fitur berkorelasi tinggi menyebabkan **multikolinearitas struktural parah** yang mendistorsi estimasi model (noise ruang fitur).  
Untuk mendapatkan subset fitur terbaik secara objektif, kita menerapkan **7 metode seleksi komprehensif** yang mencakup paradigma *Filter*, *Wrapper*, dan *Embedded*:

1. **Univariate Linear & Non-Linear (Filter):** F-Score (`f_regression`) dan Mutual Information (`mutual_info_regression`).
2. **Random Forest MDI Importance (Embedded):** Mengukur penurunan ketidakmurnian (impurity decrease) pada ansambel pohon.
3. **Permutation Importance (Model-Agnostic):** Mengukur penurunan $R^2$ saat nilai fitur diacak pada `TimeSeriesSplit`.
4. **Analisis Korelasi Pearson (Filter):** Mengidentifikasi pasangan fitur dengan korelasi linear $r > 0.8$.
5. **Variance Inflation Factor / VIF (Diagnostik):** Mendeteksi multikolinearitas simultan multivariat.
6. **Recursive Feature Elimination / RFECV (Wrapper):** Eliminasi iteratif dengan validasi silang deret waktu.
7. **LASSO L1-Regularization (Embedded):** Regularisasi penalti absolut untuk menolkan koefisien fitur yang tidak esensial.

## 10. Konsensus Voting, Redundancy Pruning, & Rekapitulasi Seleksi

Agar tidak bias pada satu metode algoritma tertentu, kita menerapkan **konsensus voting**:
1. **Peringkat Relatif:** Skor tiap metode dikonversi menjadi peringkat 1 s/d $P$ dan dirata-rata (`Avg_Rank`).
2. **Kriteria Suara (Voting):** Sebuah fitur memperoleh 1 suara jika masuk dalam Top-K metode berbasis skor (`F_score`, `MI`, `RF_Imp`, `Perm_Imp`), terpilih oleh `RFECV`, atau memiliki koefisien bukan nol pada `LASSO`.
3. **Ambang Batas Kelulusan:** Fitur dinyatakan lolos jika didukung oleh **minimal 4 dari 6 metode** (mayoritas suara).
4. **Pemangkasan Redundansi (*Redundancy Pruning*):** Dari fitur-fitur yang lolos voting, jika ada pasangan fitur yang memiliki korelasi absolut $r > 0.8$, maka fitur dengan peringkat yang lebih rendah dibuang demi mengeliminasi multikolinearitas.


```python
# Agregasi Peringkat
rank = pd.DataFrame({
    'F_score':    uni['F_score'],
    'MI':         uni['Mutual_Info'],
    'RF_Imp':     imp,
    'Perm_Imp':   perm_imp['Importance_Mean'],
    'RFE_Rank':   rfe_ranking,
    'LASSO_Coef': lasso_coef,
}).rank(ascending=False)

# RFE: ranking 1 sudah yang terbaik
rank['RFE_Rank'] = rfe_ranking.rank(ascending=True)
rank['Avg_Rank'] = rank.mean(axis=1)
rank = rank.sort_values('Avg_Rank')

# Skema Voting Mayoritas
K = max(3, len(X.columns) // 2)
MIN_VOTE = 4

vote = pd.DataFrame(index=X.columns)
for m in ['F_score', 'MI', 'RF_Imp', 'Perm_Imp']:
    vote[m] = rank[m] <= K
vote['RFECV'] = rfe_support
vote['LASSO'] = lasso_coef > 0
vote['Jumlah_Vote'] = vote.sum(axis=1)
vote['Avg_Rank'] = rank['Avg_Rank']
vote = vote.sort_values(['Jumlah_Vote', 'Avg_Rank'], ascending=[False, True])

# Filter Redundansi Korelasi (> 0.8)
lolos_vote = [c for c in rank.index if vote.loc[c, 'Jumlah_Vote'] >= MIN_VOTE]
tersisa = []
for c in lolos_vote:
    if all(fc.loc[c, t] <= 0.8 for t in tersisa):
        tersisa.append(c)

selected  = tersisa
redundant = [c for c in lolos_vote if c not in tersisa]
low_vote  = [c for c in rank.index if c not in lolos_vote]

# Visualisasi Klasemen Seleksi
fig, axes = plt.subplots(1, 2, figsize=(16, 7))

rank_display = rank.drop(columns='Avg_Rank').loc[rank.index]
sns.heatmap(rank_display, annot=True, fmt='.0f', cmap='YlOrRd_r',
            linewidths=0.5, ax=axes[0], cbar_kws={'label': 'Peringkat (1 = Terbaik)'})
axes[0].set_title('Peringkat Fitur di Setiap Metode Seleksi', fontweight='bold')
axes[0].set_xlabel('Metode Seleksi')

colors_summary = [
    'forestgreen' if f in selected else ('orange' if f in redundant else 'lightgray')
    for f in rank.sort_values('Avg_Rank').index
]
rank.sort_values('Avg_Rank')['Avg_Rank'].plot.barh(ax=axes[1], color=colors_summary)
axes[1].set_title('Rata-rata Peringkat (Hijau=Terpilih, Oranye=Redundan, Abu=Gugur)', fontweight='bold')
axes[1].set_xlabel('Avg Rank (Semakin Rendah Semakin Baik)')
axes[1].invert_xaxis()

plt.tight_layout()
plt.show()

display(Markdown(f"### Hasil Seleksi Konsensus: **{len(selected)} FITUR TERPILIH**"))
for i, f in enumerate(selected, 1):
    print(f"  {i}. {f} (Dukungan: {vote.loc[f, 'Jumlah_Vote']}/6 metode | Avg Rank: {rank.loc[f, 'Avg_Rank']:.1f})")

if len(redundant) > 0:
    display(Markdown(f"### Fitur Dieliminasi karena Redundan (Multikolinearitas r > 0.8):"))
    for f in redundant:
        print(f"  - {f} (redundan terhadap fitur terpilih berperingkat lebih tinggi)")
```

## 13. Perancangan Arsitektur Model Peramalan (Model Architecture Design)

### 13.1 Rasional Pemilihan Model Utama (Facebook/Meta Prophet)
Berdasarkan karakteristik data harga mingguan 4 varian minyak goreng nasional (Januari 2022 s/d Agustus 2026) yang telah dianalisis pada tahap *Data Preparation*, arsitektur model utama yang dirancang adalah **Prophet** (model peramalan deret waktu aditif yang dikembangkan oleh Meta/Facebook) dengan pertimbangan metodologis berikut:
1. **Dukungan Eksplisit Efek Hari Besar Keagamaan (`holidays`):** Tanggal hari besar keagamaan di Indonesia (Ramadan, Idul Fitri, Idul Adha, Nyepi, Natal) bergeser setiap tahun mengikuti kalender Lunar/Hijriah dan Saka. Prophet memodelkan efek masing-masing hari libur secara spesifik beserta jendela waktu sebelum (`lower_window`) dan sesudah (`upper_window`) hari raya.
2. **Kesesuaian untuk Ukuran Sampel Terbatas:** Dataset memiliki $\pm 241$ titik observasi mingguan per varian. Model dekomposisi struktural seperti Prophet lebih stabil dan terhindar dari *overfitting* dibandingkan arsitektur *deep learning* (LSTM/Transformer) yang membutuhkan ribuan titik data.
3. **Toleransi Native terhadap Data Hilang:** Arsitektur *continuous-time* pada Prophet mampu memproses deret waktu dengan tanggal kosong (28 Februari 2022 dan 2 Mei 2022) tanpa distorsi interpolasi.
4. **Interpretabilitas Dekomposisi Komponen:** Hasil prediksi dapat diuraikan secara transparan menjadi komponen tren jangka panjang, pola musiman tahunan, dan besaran lonjakan harga akibat hari raya keagamaan.

### 13.2 Formulasi Matematis Model
Model Prophet dirancang menggunakan spesifikasi dekomposisi aditif untuk masing-masing varian $v \in \{1, 2, 3, 4\}$:

$$y_v(t) = g_v(t) + s_v(t) + h_v(t) + \epsilon_{v,t}$$

di mana:
- **$g_v(t)$ — Komponen Tren (*Piecewise Linear Trend*):** Menangkap pergerakan harga makroekonomi jangka panjang beserta titik perubahan rezim harga (*changepoints*) $s_j$ pada waktu $t$:
  $$g(t) = \left(k + \mathbf{a}(t)^\top \boldsymbol{\delta}\right)t + \left(m + \mathbf{a}(t)^\top \boldsymbol{\gamma}\right)$$
- **$s_v(t)$ — Komponen Musiman Tahunan (*Yearly Seasonality*):** Direpresentasikan menggunakan deret Fourier orde $N=10$ untuk menangkap siklus tahunan berulang:
  $$s(t) = \sum_{n=1}^{N} \left(a_n \cos\left(\frac{2\pi nt}{365.25}\right) + b_n \sin\left(\frac{2\pi nt}{365.25}\right)\right)$$
- **$h_v(t)$ — Komponen Efek Hari Besar Keagamaan (*Holiday Effects*):** Matriks indikator regressor hari raya beserta parameter jendela waktu $[L_i, U_i]$:
  $$h(t) = \mathbf{Z}(t)\boldsymbol{\kappa}, \quad \mathbf{Z}(t) = \left[\mathbb{I}(t \in D_1), \dots, \mathbb{I}(t \in D_L)\right]$$
- **$\epsilon_{v,t} \sim \mathcal{N}(0, \sigma^2)$ — Suku Galat (*Error Term*):** Merepresentasikan fluktuasi idionsinkratik yang digunakan untuk mengestimasi batas interval ketidakpastian ($\hat{y}_{\text{lower}}, \hat{y}_{\text{upper}}$).

```python
# Spesifikasi rancangan struktur model per varian minyak goreng
arsitektur_varian = []
for idx, varian in enumerate(VARIAN_LIST, 1):
    df_v = datasets_prophet[varian]
    arsitektur_varian.append({
        'Kode Model': f'PROPHET_V{idx}',
        'Varian Target': varian,
        'Kolom Input': 'ds (datetime), y (float)',
        'Observasi Valid': len(df_v),
        'Rentang Awal': str(df_v['ds'].min().date()),
        'Rentang Akhir': str(df_v['ds'].max().date()),
        'Regressor Eksternal': f"holidays_df ({len(holidays_df)} kejadian hari besar)"
    })

df_arsitektur = pd.DataFrame(arsitektur_varian)
display(Markdown("### Tabel Spesifikasi Input & Cakupan Model per Varian"))
display(df_arsitektur)

# Ringkasan desain windowing regressor hari besar keagamaan
ringkasan_window = (
    holidays_df.groupby('holiday')[['lower_window', 'upper_window']]
    .first()
    .assign(
        Total_Kejadian=holidays_df.groupby('holiday')['ds'].count(),
        Durasi_Jendela_Efektif=lambda d: (d['upper_window'] - d['lower_window'] + 1).astype(str) + ' hari'
    )
    .reset_index()
)
display(Markdown("### Tabel Konfigurasi Jendela Waktu (*Windowing*) Regressor Hari Libur"))
display(ringkasan_window)
```

### 14.1 Konfigurasi Hyperparameter Model Prophet
Setiap model varian dirancang dengan parameter terstandarisasi berikut:

- `holidays_prior_scale = 10.0`: Memberikan ruang regularisasi yang cukup bagi model untuk menangkap efek lonjakan harga menjelang hari besar keagamaan.
- `interval_width = 0.80` (serta pengujian sensitivitas pada `0.95`): Lebar *confidence interval* untuk menentukan batas atas (*upper bound*) deteksi gejolak harga.

### 14.2 Perancangan Skema Pembagian Data & Validasi Deret Waktu
Untuk menjamin evaluasi yang bebas dari kebocoran masa depan (*future look-ahead bias*), validasi dirancang dalam dua skema komplementer:
1. **Partisi Hold-Out Kronologis (80% Data Latih : 20% Data Uji):** Memisahkan 80% observasi awal untuk pelatihan dan 20% observasi terakhir untuk menguji generalisasi prediksi *out-of-sample*.
2. **Rolling-Origin Cross-Validation (*Backtesting* Deret Waktu):** Sesuai spesifikasi proposal proyek:
   - `initial = '728 days'` ($\pm 104$ minggu / 2 tahun periode pelatihan minimum awal).
   - `period = '182 days'` ($26$ minggu jarak pergeseran antar titik *cutoff* pengujian).
   - `horizon = '84 days'` ($12$ minggu jangkauan prediksi ke depan pada setiap titik *cutoff*).

```python
# Konfigurasi Hyperparameter Terstandarisasi Prophet
PROPHET_CONFIG = {
    'growth': 'linear',
    'yearly_seasonality': True,
    'weekly_seasonality': False,
    'daily_seasonality': False,
    'seasonality_mode': 'additive',
    'changepoint_prior_scale': 0.05,
    'holidays_prior_scale': 10.0,
    'interval_width': 0.80,
}

CV_CONFIG = {
    'initial': '728 days',   # ~104 minggu (2 tahun pelatihan awal)
    'period': '182 days',    # 26 minggu jarak antar cutoff
    'horizon': '84 days',    # 12 minggu horizon peramalan
}

# Simulasi rancangan partisi Hold-Out (80:20) dan titik Cutoff Rolling-Origin CV
sample_ds = datasets_prophet['Minyak Goreng']['ds'].sort_values().reset_index(drop=True)
split_idx = int(len(sample_ds) * 0.80)
train_end_date = sample_ds.iloc[split_idx - 1]
test_start_date = sample_ds.iloc[split_idx]

max_date = sample_ds.max()
min_date = sample_ds.min()
initial_limit = min_date + pd.Timedelta(days=728)
horizon_td = pd.Timedelta(days=84)
period_td = pd.Timedelta(days=182)

cutoffs = []
cur_cutoff = max_date - horizon_td
while cur_cutoff >= initial_limit:
    cutoffs.append(cur_cutoff)
    cur_cutoff -= period_td
cutoffs = sorted(cutoffs)

print("=== RANCANGAN SKEMA PARTISI & VALIDASI DERET WAKTU ===")
print(f"1. Hold-Out Split (80:20):")
print(f"   - Data Latih (80%): {min_date.date()} s/d {train_end_date.date()} ({split_idx} minggu)")
print(f"   - Data Uji   (20%): {test_start_date.date()} s/d {max_date.date()} ({len(sample_ds) - split_idx} minggu)")
print(f"\n2. Rolling-Origin Cross-Validation ({len(cutoffs)} Titik Cutoff):")
for idx, co in enumerate(cutoffs, 1):
    print(f"   - Fold Cutoff {idx}: Pelatihan s/d {co.date()} -> Evaluasi Horizon {(co + pd.Timedelta(days=1)).date()} s/d {(co + horizon_td).date()}")

# Visualisasi Skema Rolling-Origin Cross-Validation
fig, ax = plt.subplots(figsize=(14, 5))
for idx, co in enumerate(cutoffs, 1):
    # Baris periode latih
    ax.plot([min_date, co], [idx, idx], color='#264653', linewidth=6, label='Periode Latih (Train)' if idx == 1 else "")
    # Baris periode uji (horizon 12 minggu)
    ax.plot([co, co + horizon_td], [idx, idx], color='#E63946', linewidth=6, label='Horizon Evaluasi (12 Minggu)' if idx == 1 else "")
    ax.scatter([co], [idx], color='black', s=50, zorder=5, label='Titik Cutoff' if idx == 1 else "")

ax.set_yticks(range(1, len(cutoffs) + 1))
ax.set_yticklabels([f'Cutoff {i}\n({c.strftime("%Y-%m-%d")})' for i, c in enumerate(cutoffs, 1)])
ax.set_title('Rancangan Skema Rolling-Origin Cross-Validation (Initial=104 Minggu, Period=26 Minggu, Horizon=12 Minggu)', fontweight='bold')
ax.set_xlabel('Garis Waktu Observasi (2022 - 2026)')
ax.grid(True, linestyle='--', alpha=0.4)
ax.legend(loc='lower right')
plt.tight_layout()
plt.show()
```

## 15. Perancangan Mekanisme Sistem Peringatan Dini & Instrumen Evaluasi

### 15.1 Mekanisme Deteksi Gejolak Harga (*Threshold-Based Anomaly Detection*)
Sistem peringatan dini (*early warning system*) dirancang dengan membandingkan harga aktual mingguan $y(t)$ terhadap batas atas interval kepercayaan prediksi Prophet $\hat{y}_{\text{upper}}(t)$:

$$\text{Status}(t) = \begin{cases} \text{WASPADA (Price Shock)}, & \text{jika } y(t) > \hat{y}_{\text{upper}}(t) \\ \text{NORMAL}, & \text{jika } y(t) \le \hat{y}_{\text{upper}}(t) \end{cases}$$

Besaran pelampauan ambang batas (*shock magnitude*) dihitung dalam satuan Rupiah dan persentase:
$$\Delta_{\text{shock}}(t) = \max\left(0,\; y(t) - \hat{y}_{\text{upper}}(t)\right), \quad \%\text{Shock}(t) = \frac{\Delta_{\text{shock}}(t)}{\hat{y}_{\text{upper}}(t)} \times 100\%$$

### 15.2 Instrumen & Kriteria Keberhasilan Evaluasi Model
Kinerja model pada tahap pembangunan akan diukur menggunakan 4 instrumen utama:
1. **MAPE (*Mean Absolute Percentage Error*):** $\text{MAPE} = \frac{100\%}{n} \sum_{t=1}^{n} \left|\frac{y_t - \hat{y}_t}{y_t}\right|$ dengan target keberhasilan **MAPE $< 10\%$**.
2. **RMSE (*Root Mean Squared Error*):** $\text{RMSE} = \sqrt{\frac{1}{n} \sum_{t=1}^{n} (y_t - \hat{y}_t)^2}$ untuk mengukur sensitivitas terhadap kesalahan bernominal besar.
3. **MAE (*Mean Absolute Error*):** $\text{MAE} = \frac{1}{n} \sum_{t=1}^{n} |y_t - \hat{y}_t|$ untuk mengukur rata-rata selisih absolut dalam Rupiah.
4. **Sensitivitas Deteksi Historis:** Kemampuan memicu sinyal **WASPADA** pada periode gejolak krisis minyak goreng menjelang Lebaran 2022.

```python
# Definisi fungsi-fungsi arsitektur pemodelan, evaluasi, dan deteksi dini (Model Blueprint)
def rancang_model_prophet(holidays_data, interval_width=0.80, changepoint_scale=0.05, holidays_scale=10.0):
    """
    Menginisialisasi objek model Prophet sesuai spesifikasi rancangan arsitektur.
    """
    model = Prophet(
        growth='linear',
        yearly_seasonality=True,
        weekly_seasonality=False,
        daily_seasonality=False,
        holidays=holidays_data,
        seasonality_mode='additive',
        interval_width=interval_width,
        changepoint_prior_scale=changepoint_scale,
        holidays_prior_scale=holidays_scale,
    )
    return model

def hitung_metrik_evaluasi(y_true, y_pred):
    """
    Menghitung metrik akurasi peramalan deret waktu: MAPE (%), RMSE (Rp), dan MAE (Rp).
    """
    y_true_arr = np.asarray(y_true, dtype=float)
    y_pred_arr = np.asarray(y_pred, dtype=float)
    mape_val = np.mean(np.abs((y_true_arr - y_pred_arr) / y_true_arr)) * 100.0
    rmse_val = float(np.sqrt(np.mean((y_true_arr - y_pred_arr) ** 2)))
    mae_val = float(np.mean(np.abs(y_true_arr - y_pred_arr)))
    return {'MAPE (%)': round(mape_val, 2), 'RMSE (Rp)': round(rmse_val, 2), 'MAE (Rp)': round(mae_val, 2)}

def evaluasi_status_peringatan_dini(df_actual, df_forecast):
    """
    Menggabungkan data aktual dengan hasil peramalan Prophet serta mengklasifikasikan
    status peringatan dini ('NORMAL' vs 'WASPADA') berdasarkan batas atas yhat_upper.
    """
    cols_fc = ['ds', 'yhat', 'yhat_lower', 'yhat_upper', 'trend', 'yearly', 'holidays']
    available_cols = [c for c in cols_fc if c in df_forecast.columns]
    df_eval = pd.merge(df_actual[['ds', 'y']], df_forecast[available_cols], on='ds', how='right')
    df_eval['Selisih_Upper_Rp'] = (df_eval['y'] - df_eval['yhat_upper']).clip(lower=0.0)
    df_eval['Persen_Pelampauan'] = np.where(
        df_eval['y'].notna() & (df_eval['y'] > df_eval['yhat_upper']),
        (df_eval['y'] - df_eval['yhat_upper']) / df_eval['yhat_upper'] * 100.0,
        0.0
    )
    df_eval['Status'] = np.where(
        df_eval['y'].isna(),
        'PROYEKSI',
        np.where(df_eval['y'] > df_eval['yhat_upper'], 'WASPADA', 'NORMAL')
    )
    return df_eval

# Verifikasi kesiapan blueprint fungsi perancangan model
test_blueprint = rancang_model_prophet(holidays_df, interval_width=0.80)
print("Blueprint fungsi perancangan model berhasil diverifikasi:")
print(f"  - Growth Mode           : {test_blueprint.growth}")
print(f"  - Seasonality Mode      : {test_blueprint.seasonality_mode}")
print(f"  - Yearly Seasonality    : {test_blueprint.yearly_seasonality}")
print(f"  - Interval Width Default: {test_blueprint.interval_width * 100:.0f}%")
print(f"  - Holidays Regressor    : Terintegrasi ({len(holidays_df)} baris kalender hari raya)")
```

## 16. Kesimpulan Tahap Perancangan Model

Tahap perancangan model untuk **Sistem Peringatan Dini Gejolak Harga Minyak Goreng di Indonesia** telah diselesaikan secara menyeluruh dengan rincian:
